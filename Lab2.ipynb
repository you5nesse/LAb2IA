# Partie 5 : 

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
phrases = [
    "Le football est un sport populaire",
    "Le cinéma est un art populaire",
    "Le football et le cinéma sont populaires"
]
```

```python
vectorizer = CountVectorizer()
matrice = vectorizer.fit_transform(phrases)
print(vectorizer.get_feature_names_out())
```

```python
print("\nMatrice Bag of Words :")
print(matrice.toarray())
```

# 1. 
Chaque colonne représente un mot unique extrait de l'ensemble des phrases. La valeur indique le nombre d'occurrences de ce mot dans chaque phrase.

# 2. 
- Il ne comprend pas le sens des mots.
- Il ne prend pas en compte l'ordre des mots.
- Il ne comprend pas le contexte.
- Il peut produire des matrices très grandes et creuses.
# 3. 

On perd la structure grammaticale et le sens de certaines phrases.

# Partie 6 : 


```python
from sklearn.feature_extraction.text import TfidfVectorizer

phrases = [
    "Le football est un sport populaire",
    "Le cinéma est un art populaire",
    "Le football et le cinéma sont populaires"
]
```

```python
vectorizer = TfidfVectorizer()

matrice = vectorizer.fit_transform(phrases)

print("Liste des mots :")
print(vectorizer.get_feature_names_out())
```

```python
print("\nMatrice TF-IDF :")
print(matrice.toarray().round(3))
```

# 1. 
- BoW : compte le nombre d'apparitions de chaque mot dans un document.
- TF-IDF : attribue un poids aux mots selon leur fréquence dans le document et leur rareté dans l'ensemble des documents.
# 2.

Parce qu'ils apparaissent dans plusieurs documents et ne permettent pas de distinguer facilement un document des autres.

Exemple : les mots « le », « la », « et » sont très fréquents et apportent généralement peu d'informations.
